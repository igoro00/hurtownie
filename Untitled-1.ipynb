```python
import pandas as pd

df = pd.read_csv('SaaS-Sales.csv')
```

```python
df
```

```python
# convert Order Date to datetime, sort, aggregate profit per day and plot
df_sorted = df.copy()
df_sorted['Order Date'] = pd.to_datetime(df_sorted['Order Date'])
df_sorted = df_sorted.sort_values('Order Date')

profit_daily = df_sorted.groupby('Order Date')['Profit'].sum()

ax = profit_daily.plot(figsize=(12,5), title='Profit over time')
ax.set_xlabel('Date')
ax.set_ylabel('Profit')
ax.grid(True)
```
